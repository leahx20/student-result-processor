```python
#creating class which is the student
import grading_utils


class  Student:
    def __init__(self, name, registration_number, marks):
        self.name = name
        self.registration_number = registration_number
        self.marks = marks
        self.average = 0
        self.result = ""

    def calculate_average(self):
        self.average = sum(self.marks) / len(self.marks)
        return self.average

    def classify_result(self):
        if self.average >= 70:
            self.result = "Distinction"
        elif self.average >= 50:
            self.result = "Pass"
        else:
            self.result = "Fail"

        return self.result

    def display_result(self):
        print("================================")
        print("       STUDENT RESULT")
        print("================================")
        print(f"Name: {self.name}")
        print(f"Registration No: {self.registration_number}")
        print(f"Marks: {self.marks}")
        print(f"Average: {self.average:.2f}")
        print(f"Result: {self.result}")
        print("================================")
```

created our class which is student

```python
#creating the object
student1 = Student(
    "Leah Mungathia",
    "251568DIT",
    [75, 69, 84]
)

student1.calculate_average()
student1.classify_result()
student1.display_result()
```
